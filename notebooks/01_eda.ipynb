## Key findings
- 6.7% of applicants are delinquent: heavy class imbalance, so accuracy is not a useful metric.
- Default rate falls steadily with age (11.9% at 18-29 vs 2.3% at 70+).
- Utilization is the strongest non-history signal: 1.8% below 10% utilization vs 37.3% above 100%.
- Any 90-day lateness is a huge flag: 4.7% with none vs 33.7% with one.
- Debt ratio is almost uninformative on its own (large values are mostly missing-income artefacts).
